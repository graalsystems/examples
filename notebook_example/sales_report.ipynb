# Monthly sales report

The cell tagged `parameters` holds the defaults; the job's `options.parameters` override them.

```python
# Defaults, overridden by the job's options.parameters
region = "all"
min_quantity = 0
```

```python
import pandas as pd

sales = pd.DataFrame(
    [
        ("Bretagne", "Rennes", "2026-01", "apple", 120, 1.25),
        ("Bretagne", "Brest", "2026-01", "pear", 80, 2.10),
        ("Bretagne", "Rennes", "2026-02", "apple", 95, 1.30),
        ("Occitanie", "Toulouse", "2026-01", "apple", 150, 1.20),
        ("Occitanie", "Montpellier", "2026-02", "pear", 60, 2.40),
        ("Occitanie", "Toulouse", "2026-02", "pear", 70, 2.35),
        ("Ile-de-France", "Paris", "2026-01", "apple", 300, 1.45),
        ("Ile-de-France", "Paris", "2026-02", "pear", 210, 2.60),
    ],
    columns=["region", "city", "month", "product", "quantity", "unit_price"],
)
sales["revenue"] = sales["quantity"] * sales["unit_price"]
if region != "all":
    sales = sales[sales["region"] == region]
sales = sales[sales["quantity"] >= min_quantity]
assert not sales.empty, f"no sale left for region={region!r}, min_quantity={min_quantity}"
print(f"{len(sales)} sales lines (region={region}, min_quantity={min_quantity})")
```

```python
report = (
    sales.groupby(["region", "month"], as_index=False)[["quantity", "revenue"]]
    .sum()
    .sort_values(["region", "month"])
)
report["revenue"] = report["revenue"].round(2)
print(report.to_string(index=False))
```

```python
total = round(float(sales["revenue"].sum()), 2)
print(f"TOTAL REVENUE {total}")
```
